# QWGNN Training Configuration

This JSON dictionary contains the main hyperparameters and settings for training the **Quantum Walk Graph Neural Network (QWGNN)** on the `PROTEINS_SMALL` dataset.

| Key | Value | Description |
|-----|-------|-------------|
| `mode` | `"train"` | Specifies the execution mode. Options: `"train"` or `"test"` |
| `dataset_name` | `"PROTEINS_SMALL"` | Name of the dataset used for training |
| `dataset_node` | `6` | Maximum number of nodes in the graphs to consider |
| `batch_size` | `64` | Number of graphs per training batch |
| `model` | `"QWGNN"` | Specifies the model architecture to use |
| `lr` | `0.001` | Learning rate for the optimizer |
| `dropout` | `0.2` | Dropout rate applied in the network for regularization |
| `weight_decay` | `0.0005` | Weight decay (L2 regularization) applied in the optimizer |
| `num_epochs` | `300` | Total number of training epochs |
| `save_freq` | `20` | Frequency (in epochs) to save model checkpoints |
| `save_path` | `"./runs/PROTIENS/QWGNN"` | Directory path to save logs and model checkpoints |
| `checkpoint_path` | `None` | Path to a pre-trained model checkpoint (if any) |
| `patience` | `10` | Number of epochs with no improvement before early stopping is triggered |

### Notes
- The `dataset_node` parameter indicates that only graphs with **up to 6 nodes** will be used.  
- The `patience` parameter is relevant if early stopping is implemented during training.  
- The `save_path` directory will store both **logs** and **model checkpoints**, useful for monitoring and resuming training.
